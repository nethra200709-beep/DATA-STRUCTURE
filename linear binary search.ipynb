```python
def linearsearch(arr, key):
    for i in range(len(arr)):
        if arr[i] == key:
            return i
    return -1


def binarysearch(arr, key):
    low = 0
    high = len(arr) - 1

    while low <= high:
        mid = (low + high) // 2

        if arr[mid] == key:
            return mid
        elif arr[mid] < key:
            low = mid + 1
        else:
            high = mid - 1

    return -1


arr = list(map(int, input("Enter elements separated by space: ").split()))
key = int(input("Enter element to search: "))

position = linearsearch(arr, key)

if position != -1:
    print("Linear Search: Element found at position", position)
else:
    print("Linear Search: Element not found")

print("Binary Search")
arr.sort()
print("Sorted array:",arr)
position=binarysearch(arr,key)

if position!=-1:
    print("Binary search:element found")
else:
    print("Binary search:element not found")
```
